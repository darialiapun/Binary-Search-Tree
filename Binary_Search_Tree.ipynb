```python
class Node: # тип Node - узел
    def __init__(self, key): #создаем (узел, число в нем)
        self.key = key # записываем число внутрь узла
        self.left = None # в нижнем левом и правм узле пока ничего нет
        self.right = None


class BST: # двоичное дерево поиска
    def __init__(self):
        self.root = None # пока у дерева есть только корень и он пустой

    # вставка: идём влево если меньше, вправо если больше
    def insert(self, key): # вставка
        self.root = self._insert(self.root, key) #передаем корень дерева и число которое вставляем

    def _insert(self, node, key): # название дерева, node(узел) это наш корень, и само число
        if node is None:
            return Node(key) # если узла нет то создаем новый и вставляем значение
        if key < node.key:
            node.left = self._insert(node.left, key) # если вставляемое число меньше числа в текущем узле, нужно идти влево
        elif key > node.key:
            node.right = self._insert(node.right, key)
        return node  # дубликаты не добавляем

    # поиск циклом
    def search(self, key):
        node = self.root
        while node is not None:
            if key == node.key:
                return True
            if key < node.key:
                node = node.left
            else:
                node = node.right
        return False

    # удаление
    def delete(self, key):
        self.root = self._delete(self.root, key)

    def _delete(self, node, key):
        if node is None:
            return None
        if key < node.key:
            node.left = self._delete(node.left, key)
        elif key > node.key:
            node.right = self._delete(node.right, key)
        else:
            if node.left is None:
                return node.right
            if node.right is None:
                return node.left
            m = node.right
            while m.left is not None:
                m = m.left
            node.key = m.key
            node.right = self._delete(node.right, m.key)
        return node

    # обходы
    def inorder(self):
        res = []
        def walk(node):
            if node is None:
                return
            walk(node.left)
            res.append(node.key)
            walk(node.right)
        walk(self.root)
        return res

    def preorder(self):
        res = []
        def walk(node):
            if node is None:
                return
            res.append(node.key)
            walk(node.left)
            walk(node.right)
        walk(self.root)
        return res

    def postorder(self):
        res = []
        def walk(node):
            if node is None:
                return
            walk(node.left)
            walk(node.right)
            res.append(node.key)
        walk(self.root)
        return res

    def find_min(self):
        if self.root is None:
            return None
        node = self.root
        while node.left is not None:
            node = node.left
        return node.key

    def find_max(self):
        if self.root is None:
            return None
        node = self.root
        while node.right is not None:
            node = node.right
        return node.key

    # высота: пустое дерево 0, один узел 1
    def height(self):
        def h(node):
            if node is None:
                return 0
            return 1 + max(h(node.left), h(node.right))
        return h(self.root)

    # сбалансировано, если в каждом узле высоты поддеревьев отличаются не больше чем на 1
    def is_balanced(self):
        def h(node):
            if node is None:
                return 0
            return 1 + max(h(node.left), h(node.right))

        def check(node):
            if node is None:
                return True
            if abs(h(node.left) - h(node.right)) > 1:
                return False
            return check(node.left) and check(node.right)

        return check(self.root)


# тесты

def make_tree(keys):
    t = BST()
    for k in keys:
        t.insert(k)
    return t

KEYS = [50, 30, 70, 20, 40, 60, 80]


def test_insert_search():
    t = make_tree(KEYS)
    for k in KEYS:
        assert t.search(k)
    assert not t.search(45)
    assert not t.search(100)
    assert not BST().search(1)          # пустое дерево
    t.insert(50)                        # дубликат
    assert t.inorder() == [20, 30, 40, 50, 60, 70, 80]


def test_delete():
    # случай 1: лист
    t = make_tree(KEYS)
    t.delete(20)
    assert t.inorder() == [30, 40, 50, 60, 70, 80]
    # случай 2: один ребёнок
    t = make_tree([50, 30, 40])
    t.delete(30)
    assert t.inorder() == [40, 50]
    # случай 3: два ребёнка
    t = make_tree([50, 30, 70, 60, 80])
    t.delete(70)
    assert t.inorder() == [30, 50, 60, 80]
    # удаление корня
    t = make_tree(KEYS)
    t.delete(50)
    assert t.inorder() == [20, 30, 40, 60, 70, 80]
    assert t.root.key == 60
    # нет такого ключа
    t = make_tree(KEYS)
    t.delete(999)
    assert t.inorder() == [20, 30, 40, 50, 60, 70, 80]
    # удалить всё
    for k in KEYS:
        t.delete(k)
    assert t.root is None


def test_traversals():
    t = make_tree(KEYS)
    assert t.inorder() == [20, 30, 40, 50, 60, 70, 80]
    assert t.preorder() == [50, 30, 20, 40, 70, 60, 80]
    assert t.postorder() == [20, 40, 30, 60, 80, 70, 50]
    assert BST().inorder() == []
    assert make_tree([7]).preorder() == [7]


def test_min_max():
    t = make_tree(KEYS)
    assert t.find_min() == 20
    assert t.find_max() == 80
    assert BST().find_min() is None
    assert BST().find_max() is None


def test_height_balanced():
    assert make_tree(KEYS).height() == 3
    assert BST().height() == 0
    assert make_tree([1]).height() == 1
    assert make_tree(KEYS).is_balanced()
    assert BST().is_balanced()
    assert make_tree([50, 30, 70, 20]).is_balanced()      # разница 1 — ещё можно
    assert not make_tree([50, 30, 20]).is_balanced()      # разница 2
    t = make_tree(range(1, 11))
    assert t.height() == 10
    assert not t.is_balanced()


def test_random():
    import random
    for _ in range(50):
        keys = random.sample(range(1000), 30)
        t = make_tree(keys)
        assert t.inorder() == sorted(keys)
        for k in keys[:10]:
            t.delete(k)
        assert t.inorder() == sorted(keys[10:])


if __name__ == '__main__':
    test_insert_search()
    test_delete()
    test_traversals()
    test_min_max()
    test_height_balanced()
    test_random()
    print('Все тесты пройдены')
```
