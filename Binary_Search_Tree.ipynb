```python
class Node: # тип node - узел
    def __init__(self, key): #создаем (узел, число в нем)
        self.key = key # записываем число внутрь узла
        self.left = None # в нижнем левом и правм узле пока ничего нет
        self.right = None


class BST: # двоичное дерево поиска
    def __init__(self):
        self.root = None # пока у дерева есть только корень и он пустой

    # вставка
    def insert(self, key): # вставка
        self.root = self._insert(self.root, key) #передаем корень дерева и число которое вставляем

    def _insert(self, node, key): # название дерева, node(узел) это наш корень, и само число
        if node is None:
            return Node(key) # если узла нет то создаем новый и вставляем значение
        if key < node.key:
            node.left = self._insert(node.left, key) # если вставляемое число меньше числа в текущем узле, нужно идти влево
        elif key > node.key:
            node.right = self._insert(node.right, key)
        return node  # дубликаты не добавляем

    # поиск циклом
    def search(self, key):
        node = self.root # начинаем поиск с корня, node - узел в котором мы сейчас
        while node is not None:
            if key == node.key:
                return True # если нашли число сразу по выходим
            if key < node.key: # если не нашли двигаемся вправо влево
                node = node.left
            else:
                node = node.right
        return False # если числа нет в целом

    # удаление
    def delete(self, key): # начинаем поиск элемента который нужно удалить с корня, и затем если корень поменялся присваиваем ему новое значение
        self.root = self._delete(self.root, key)

    def _delete(self, node, key):
        if node is None: # случай если один ребенок или нет детей
            return None # если по ходу поиска дошли до пустого места, то числа уже нет в дрееве и удалять уже нечего
        if key < node.key:
            node.left = self._delete(node.left, key) # ищем нужное число также как в search двигаемся влево вправо
        elif key > node.key:
            node.right = self._delete(node.right, key) # присваивание на место удаленного числа следующего числа
        else: # случай если два ребенка
            if node.left is None:
                return node.right
            if node.right is None:
                return node.left
            m = node.right # чтобы убрать число из корня, мы должны заменить его на минимальное число из правой части
            while m.left is not None:
                m = m.left # идем от узла в правую часть и вней берем самое левое число, те самое маленькое
            node.key = m.key # это самое маленькое число теперь корень
            node.right = self._delete(node.right, m.key) # удаляем поднятое в корнеь число из правой части
        return node

    # обходы, способ пройти по всему дереву и записать числа в список
    def inorder(self): # первый способ узел записываетс я в середине
        res = [] # пустой спиосок
        def walk(node): # начинаем с узла node
            if node is None:
                return # если его нет выходим
            walk(node.left) # обходим всю левую часть
            res.append(node.key) # записываем сам узел в конец существующего на данный момент списка с числами из левой части
            walk(node.right) # обходим всю правую часть
        walk(self.root)
        return res

    def preorder(self): # узел записываеся сначала
        res = []
        def walk(node):
            if node is None:
                return
            res.append(node.key)
            walk(node.left)
            walk(node.right)
        walk(self.root)
        return res

    def postorder(self): # узел записывается в конце, те корень в конце списка
        res = []
        def walk(node):
            if node is None:
                return
            walk(node.left)
            walk(node.right)
            res.append(node.key)
        walk(self.root)
        return res

    def find_min(self):  # поиск минимума
        if self.root is None:
            return None # если дерево пустое, минимума нет, возвращаем none
        node = self.root
        while node.left is not None: # самое маленьке число в левой части самое левое
            node = node.left
        return node.key

    def find_max(self): # все тоже самое только с максимумом
        if self.root is None:
            return None
        node = self.root
        while node.right is not None:
            node = node.right
        return node.key

    # высота(сколько у дерева "этажей"): пустое дерево 0, один узел 1
    def height(self):
        def h(node):
            if node is None:
                return 0 # у пустого дерева 0
            return 1 + max(h(node.left), h(node.right)) # у узла высота равна 1 (сам узел) + высота более высокой из двух частей max(a, b) выбирает большее из двух чисел.
        return h(self.root)

    # сбалансировано (чтобы дерево было более эффективным чем список), если в каждом узле высоты поддеревьев отличаются не больше чем на 1
    def is_balanced(self):
        def h(node):  # считаем количество этажей
            if node is None:
                return 0
            return 1 + max(h(node.left), h(node.right))

        def check(node):
            if node is None:
                return True
            if abs(h(node.left) - h(node.right)) > 1: # сравниваем высоту левой и правой части
                return False # если высоты отличаются больше чем на 1, дерево перекошено, возвращаем False
            return check(node.left) and check(node.right)  # проверяем также все узлы слева и справа

        return check(self.root) # запускаем проверку с корня


# тесты

def make_tree(keys): # создаем пустое дерево и вставляем в него все числа из списка
    t = BST()
    for k in keys:
        t.insert(k)
    return t

KEYS = [50, 30, 70, 20, 40, 60, 80]


def test_insert_search():
    t = make_tree(KEYS)
    for k in KEYS:
        assert t.search(k) # подтверждение что условие верное, проверка что каждое вставленное число находится
    assert not t.search(45) # проверяем что числа 45 нет в дереве
    assert not t.search(100)
    assert not BST().search(1)          # пустое дерево
    t.insert(50)                        # дубликат
    assert t.inorder() == [20, 30, 40, 50, 60, 70, 80]


def test_delete(): # проверяем все три случая удаления, удаление корня, удаление несуществующего числа и удаление всех чисел (в конце корень должен стать None)
    # случай 1: лист
    t = make_tree(KEYS)
    t.delete(20)
    assert t.inorder() == [30, 40, 50, 60, 70, 80]
    # случай 2: один ребёнок
    t = make_tree([50, 30, 40])
    t.delete(30)
    assert t.inorder() == [40, 50]
    # случай 3: два ребёнка
    t = make_tree([50, 30, 70, 60, 80])
    t.delete(70)
    assert t.inorder() == [30, 50, 60, 80]
    # удаление корня
    t = make_tree(KEYS)
    t.delete(50)
    assert t.inorder() == [20, 30, 40, 60, 70, 80]
    assert t.root.key == 60
    # нет такого ключа
    t = make_tree(KEYS)
    t.delete(999)
    assert t.inorder() == [20, 30, 40, 50, 60, 70, 80]
    # удалить всё
    for k in KEYS:
        t.delete(k)
    assert t.root is None


def test_traversals(): # проверяет, что три обхода дают правильные списки
    t = make_tree(KEYS)
    assert t.inorder() == [20, 30, 40, 50, 60, 70, 80]
    assert t.preorder() == [50, 30, 20, 40, 70, 60, 80]
    assert t.postorder() == [20, 40, 30, 60, 80, 70, 50]
    assert BST().inorder() == []
    assert make_tree([7]).preorder() == [7]


def test_min_max(): # проверяет минимум и максимум, а также пустое дерево
    t = make_tree(KEYS)
    assert t.find_min() == 20
    assert t.find_max() == 80
    assert BST().find_min() is None
    assert BST().find_max() is None


def test_height_balanced(): # проверяет высоту и баланс
    assert make_tree(KEYS).height() == 3
    assert BST().height() == 0
    assert make_tree([1]).height() == 1
    assert make_tree(KEYS).is_balanced()
    assert BST().is_balanced()
    assert make_tree([50, 30, 70, 20]).is_balanced()      # разница 1 — ещё можно
    assert not make_tree([50, 30, 20]).is_balanced()      # разница 2
    t = make_tree(range(1, 11))
    assert t.height() == 10
    assert not t.is_balanced()


def test_random(): # подключает модуль случайных чисел
    import random
    for _ in range(50):
        keys = random.sample(range(1000), 30) # берёт 30 разных случайных чисел от 0 до 999
        t = make_tree(keys)
        assert t.inorder() == sorted(keys)
        for k in keys[:10]:
            t.delete(k) # удаляем первые 10 и проверяем, что остались ровно остальные и по порядку
        assert t.inorder() == sorted(keys[10:])


if __name__ == '__main__':
    test_insert_search()
    test_delete()
    test_traversals()
    test_min_max()
    test_height_balanced()
    test_random()
    print('Все тесты пройдены')
```
